GOOD LOAN METRICS

```python
df["loan_status"].value_counts( )
```

```python
good_loan_statuses=["Fully Paid","Current"]
good_loan=df[
    df["loan_status"].isin(good_loan_statuses)
]
good_loan_applications=good_loan['id'].count( )
total_loan_applications=df["id"].count( )
good_loan_percentage=(good_loan_applications/total_loan_applications)*100
good_loan_funded_amount=good_loan["loan_amount"].sum( )
good_loan_funded_amount_million=good_loan_funded_amount/1000000
good_Loan_total_received=good_loan["total_payment"].sum( )
good_loan_total_received_million=good_Loan_total_received/1000000
print("Good Loan Applications:",good_loan_applications)
print(f"Good Loan Percentage:{good_loan_percentage:.2f}%")
print("Good Loan Funded Amount:${:.2f}M".format(good_loan_funded_amount_million))
print(f"Good Loan Total Received:${good_loan_total_received_million:.2f}M")
```

BAD LOAN METRICS

```python
bad_loan_statuses=["Charged Off"]
bad_loan=df[
    df["loan_status"].isin(bad_loan_statuses)
]
bad_loan_applications=bad_loan['id'].count( )
total_loan_applications=df["id"].count( )
bad_loan_percentage=(bad_loan_applications/total_loan_applications)*100
bad_loan_funded_amount=bad_loan["loan_amount"].sum( )
bad_loan_funded_amount_million=bad_loan_funded_amount/1000000
bad_Loan_total_received=bad_loan["total_payment"].sum( )
bad_loan_total_received_million=bad_Loan_total_received/1000000
print("Bad Loan Applications:",bad_loan_applications)
print(f"Bad Loan Percentage:{bad_loan_percentage:.2f}%")
print("Bad Loan Funded Amount:${:.2f}M".format(bad_loan_funded_amount_million))
print(f"Bad Loan Total Received:${bad_loan_total_received_million:.2f}M")
```

MONTHLY TRENDS BY ISSUE DATE FOR TOTAL FUNDED AMOUNT

```python
# Monthly Trends for Total Funded Amount
monthly_funded_amount = df.groupby(
    df["issue_date"].dt.to_period("M")
)["loan_amount"].sum()
print(monthly_funded_amount)
```

```python
monthly_funded_amount_million = monthly_funded_amount / 1000000
print(monthly_funded_amount_million.round(2))
```

```python

import matplotlib.pyplot as plt
monthly_received = (
    df.sort_values("issue_date")
    .assign(
        month_name=lambda x: x["issue_date"].dt.strftime("%b %Y")
    )
    .groupby("month_name", sort=False)["total_payment"]
    .sum()
    .div(1000000)
    .reset_index(name="amount_received_millions")
)
plt.figure(figsize=(12, 5))

plt.fill_between(
    range(len(monthly_received)),
    monthly_received["amount_received_millions"],
    alpha=0.4
)

plt.plot(
    range(len(monthly_received)),
    monthly_received["amount_received_millions"],
    marker="o",
    linewidth=2
)
for i, row in monthly_received.iterrows():
    plt.text(
        i,
        row["amount_received_millions"] + 0.1,
        f'{row["amount_received_millions"]:.2f}',
        ha="center",
        fontsize=8
    )

plt.title("Monthly Total Amount Received", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Amount Received ($ Millions)")
plt.xticks(
    range(len(monthly_received)),
    monthly_received["month_name"],
    rotation=45
)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

MONTHLY TRENDS BY ISSUE DATE FOR TOTAL FUNDED AMOUNT

```python

import matplotlib.pyplot as plt
monthly_funded = (
    df.sort_values("issue_date")
    .assign(
        month_name=lambda x: x["issue_date"].dt.strftime("%b %Y")
    )
    .groupby("month_name", sort=False)["loan_amount"]
    .sum()
    .div(1000000)
    .reset_index(name="loan_amount_millions")
)

plt.figure(figsize=(12, 5))

plt.fill_between(
    range(len(monthly_funded)),
    monthly_funded["loan_amount_millions"],
    alpha=0.4
)

plt.plot(
    range(len(monthly_funded)),
    monthly_funded["loan_amount_millions"],
    marker="o",
    linewidth=2
)

for i, row in monthly_funded.iterrows():
    plt.text(
        i,
        row["loan_amount_millions"] + 0.1,
        f'{row["loan_amount_millions"]:.2f}',
        ha="center",
        fontsize=8
    )

plt.title("Monthly Total Funded Amount", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Funded Amount ($ Millions)")
plt.xticks(
    range(len(monthly_funded)),
    monthly_funded["month_name"],
    rotation=45
)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

MONTHLY TRENDS BY ISSUE DATE FOR TOTAL LOAN APPLICATIONS

```python

import matplotlib.pyplot as plt
monthly_applications = (
    df.sort_values("issue_date")
    .assign(
        month_name=lambda x: x["issue_date"].dt.strftime("%b %Y")
    )
    .groupby("month_name", sort=False)["id"]
    .count()
    .reset_index(name="loan_applications")
)
plt.figure(figsize=(12, 5))

plt.fill_between(
    range(len(monthly_applications)),
    monthly_applications["loan_applications"],
    alpha=0.4
)

plt.plot(
    range(len(monthly_applications)),
    monthly_applications["loan_applications"],
    marker="o",
    linewidth=2
)

for i, row in monthly_applications.iterrows():
    plt.text(
        i,
        row["loan_applications"] + 5,
        f'{row["loan_applications"]:.0f}',
        ha="center",
        fontsize=8
    )

plt.title("Monthly Total Loan Applications", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Number of Applications")
plt.xticks(
    range(len(monthly_applications)),
    monthly_applications["month_name"],
    rotation=45
)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

REGIONAL ANALYSIS BY STATE FOR TOTAL FUNDED AMOUNT

```python
print(df.columns.tolist())
```

```python

import matplotlib.pyplot as plt
state_funded = (
    df.groupby("address_state")["loan_amount"]
    .sum()
    .sort_values(ascending=False)
    .div(1000000)
)
plt.figure(figsize=(12, 6))

plt.bar(
    state_funded.index,
    state_funded.values
)
for i, value in enumerate(state_funded.values):
    plt.text(
        i,
        value,
        f"{value:.1f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

plt.title("Total Funded Amount by State", fontsize=14)
plt.xlabel("State")
plt.ylabel("Funded Amount ($ Millions)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

LONG TERM ANALYSIS BY TOTAL FUNDED AMOUNT

```python

import matplotlib.pyplot as plt
yearly_funded = (
    df.groupby(df["issue_date"].dt.year)["loan_amount"]
    .sum()
    .div(1000000)
)
plt.figure(figsize=(10, 5))

plt.plot(
    yearly_funded.index,
    yearly_funded.values,
    marker="o",
    linewidth=2
)
for year, amount in yearly_funded.items():
    plt.text(
        year,
        amount,
        f"{amount:.2f}",
        ha="center",
        va="bottom"
    )

plt.title("Long-Term Analysis - Total Funded Amount by Year", fontsize=14)
plt.xlabel("Year")
plt.ylabel("Total Funded Amount ($ Millions)")
plt.xticks(yearly_funded.index)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
